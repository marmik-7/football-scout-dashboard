```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)

PROJECT_ROOT = Path("..")

DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "players_engineered.csv"
)

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
print("Number of records:", len(df))
print("Unique player names:", df["Player"].nunique())

df.head()
```

```python
SIMILARITY_FEATURES = [
    "Finishing_Score",
    "Creativity_Score",
    "Passing_Score",
    "Progression_Score",
    "Defending_Score",
    "BallSecurity_Score",
    "Aerial_Score"
]

POSITION_COLUMN = "Pos"

print("Similarity features:", len(SIMILARITY_FEATURES))

print("\nMissing values:")
print(df[SIMILARITY_FEATURES].isna().sum())

print("\nPosition distribution:")
print(df[POSITION_COLUMN].value_counts())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
df_scaled = df.copy()

scalers = {}

for position in df[POSITION_COLUMN].unique():

    position_mask = df[POSITION_COLUMN] == position

    scaler = StandardScaler()

    df_scaled.loc[position_mask, SIMILARITY_FEATURES] = (
        scaler.fit_transform(
            df.loc[position_mask, SIMILARITY_FEATURES]
        )
    )

    scalers[position] = scaler

print("Standardization completed.")

for position in scalers:
    print(f"{position}: {len(df[df['Pos'] == position])} players")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def find_similar_players(player_name, position=None, top_n=5):

    # Find the selected player
    player_matches = df[df["Player"] == player_name]

    if player_matches.empty:
        raise ValueError(f"Player not found: {player_name}")

    if len(player_matches) > 1:
        raise ValueError(
            "Multiple players share this name. "
            "Use a unique player identifier."
        )

    player = player_matches.iloc[0]

    # Use the player's position
    player_position = player["Pos"]

    if position is not None and position != player_position:
        raise ValueError("Selected position does not match player position.")

    # Get players from the same position
    candidates = df_scaled[
        df_scaled["Pos"] == player_position
    ].copy()

    # Find the selected player's standardized vector
    player_index = player.name

    player_vector = df_scaled.loc[
        player_index, SIMILARITY_FEATURES
    ].to_numpy().reshape(1, -1)

    # Calculate similarity against candidates
    candidate_vectors = candidates[
        SIMILARITY_FEATURES
    ].to_numpy()

    similarities = cosine_similarity(
        player_vector,
        candidate_vectors
    )[0]

    # Add similarity scores
    candidates["Similarity"] = similarities

    # Exclude the selected player
    candidates = candidates[
        candidates.index != player_index
    ]

    # Return top similar players
    results = candidates.nlargest(
        top_n, "Similarity"
    )

    # Retrieve original attribute scores
    original_results = df.loc[
        results.index
    ].copy()

    original_results["Similarity"] = (
        results["Similarity"]
    )

    return original_results[
        [
            "Player", "Nation", "Squad", "Pos",
            "Similarity"
        ] + SIMILARITY_FEATURES
    ].reset_index(drop=True)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def find_similar_players_raw(player_name, top_n=5):

    player_matches = df[df["Player"] == player_name]

    if len(player_matches) != 1:
        raise ValueError("Player name must uniquely identify one player.")

    player = player_matches.iloc[0]
    player_position = player["Pos"]

    candidates = df[df["Pos"] == player_position].copy()

    player_vector = player[SIMILARITY_FEATURES].to_numpy().reshape(1, -1)
    candidate_vectors = candidates[SIMILARITY_FEATURES].to_numpy()

    similarities = cosine_similarity(
        player_vector, candidate_vectors
    )[0]

    candidates["Similarity"] = similarities
    candidates = candidates[candidates.index != player.name]

    results = candidates.nlargest(top_n, "Similarity")

    return results[
        ["Player", "Nation", "Squad", "Pos", "Similarity"]
        + SIMILARITY_FEATURES
    ].reset_index(drop=True)
```

```python
find_similar_players_raw("Erling Haaland", top_n=5)
```

## Similarity Method Selection

Two approaches were evaluated:

1. Position-wise Standardized Cosine Similarity
2. Raw Attribute Cosine Similarity

Position-wise standardization was selected as the primary approach because it
accounts for differences in attribute distributions within each position group.

The raw approach was retained as an experimental baseline.

Similarity scores represent profile resemblance and should not be interpreted
as probabilities or percentages of player equivalence.
