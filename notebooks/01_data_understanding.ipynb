```python
missing = (
    df.isna()
      .sum()
      .sort_values(ascending=False)
)

missing.head(20)
```

```python
missing_percent = (
    df.isna()
      .mean()
      * 100
).sort_values(ascending=False)

missing_percent.head(20)
```

```python
unique_players = df["Player"].nunique()

print("Unique players:", unique_players)
print("Repeated records:", len(df) - unique_players)
```

```python
player_counts = (
    df["Player"]
    .value_counts()
    .reset_index()
)

player_counts.columns = ["Player", "Count"]

duplicate_players = player_counts[player_counts["Count"] > 1]

print("Players appearing multiple times:")
display(duplicate_players.head(20))
```

```python
player_team_duplicates = (
    df.groupby(["Player", "Squad"])
    .size()
    .reset_index(name="Count")
)

real_duplicates = player_team_duplicates[
    player_team_duplicates["Count"] > 1
]

print("Duplicate Player + Squad rows:")
display(real_duplicates.head(20))
```

## Observations

- The dataset contains **2,854 player-season records** and **165 features**.
- Each row represents a player's statistics for a specific club and competition during the 2024–25 season.
- There are **2,702 unique players**, indicating that some players appear multiple times due to transfers or appearances for multiple clubs.
- No duplicate **Player + Squad** records were found, confirming that duplicate player names are legitimate records rather than duplicate entries.
- Goalkeeper-related statistics contain many missing values because they are only applicable to goalkeepers.
- Several statistics are duplicated across different FBref tables (e.g., `Ast` vs `Ast_stats_passing`, `PrgP` vs `PrgP_stats_passing`), requiring feature selection.
- The `Pos` column contains multi-position players (e.g., DF,DF-MF, FW-MF), so position standardization will be required.
- Players with very low minutes may not provide reliable scouting profiles and may need to be filtered before analysis.
- Some features represent totals, while others represent per-90 statistics or percentages. These should not be mixed directly when computing player similarity.
- The dataset is well suited for scouting analysis but requires preprocessing before feature engineering and machine learning.
